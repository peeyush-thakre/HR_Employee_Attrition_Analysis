## 14. Correlation Analysis

For numerical columns:

## 15. Create Attrition Rate


For HR analysis, one of the most useful metrics is:

Attrition Rate = Employees who left / Total Employees × 100

```python
attrition_rate = (
    df["Attrition"].eq("Yes").mean() * 100
)

print(f"Attrition Rate: {attrition_rate:.2f}%")
```

## 16. Analyze Attrition by Multiple Factors

For example:

**Department + Attrition**

```python
dept_attrition = pd.crosstab(
    df["Department"],
    df["Attrition"],
    normalize="index"
) * 100

dept_attrition
```

**Job Role + Attrition**

```python
role_attrition = pd.crosstab(
    df["JobRole"],
    df["Attrition"],
    normalize="index"
) * 100

role_attrition
```

**Overtime + Job Satisfaction + Attrition**

```python
pd.crosstab(
    [df["OverTime"], df["JobSatisfaction"]],
    df["Attrition"],
    normalize="index"
) * 100
```

This is multivariate analysis.
